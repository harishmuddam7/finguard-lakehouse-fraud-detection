```python
import time
import json
import random
from datetime import datetime

# Volume landing target for Auto Loader
landing_dir = "/Volumes/aegis_fraud_workspace/finguard/finguard_volume/raw_landing"

# Active customer IDs from dim_customers
users = [f"USR_{i:03d}" for i in range(1, 11)]
cities = ["Hyderabad", "Bengaluru", "Mumbai", "London", "Dubai", "Singapore"]
merchants = ["Amazon Web Services", "Apple Store", "Starbucks", "Delta Airlines", "Zara Fashion", "Shell Express"]

print(f"[PRODUCER] Writing synthetic streaming events to: {landing_dir}")

# Emit 10 micro-batches of transactions
for batch_num in range(1, 11):
    batch_events = []
    
    # Generate 3 to 6 transactions per payload file
    for _ in range(random.randint(3, 6)):
        user = random.choice(users)
        
        # Inject ~25% anomalies (high spend, cross-border locations)
        trigger_anomaly = random.random() < 0.25
        if trigger_anomaly:
            amount = round(random.uniform(3200.0, 9800.0), 2)
            location = random.choice(["Dubai", "London", "Singapore"])
        else:
            amount = round(random.uniform(15.0, 320.0), 2)
            location = "Hyderabad"
            
        event = {
            "transaction_id": f"TXN_{random.randint(1000000, 9999999)}",
            "user_id": user,
            "amount": amount,
            "merchant": random.choice(merchants),
            "location": location,
            "timestamp": datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S")
        }
        batch_events.append(event)
        
    target_file = f"{landing_dir}/payload_batch_{batch_num}_{int(time.time())}.json"
    
    # Direct POSIX file write into Unity Catalog Volume
    with open(target_file, "w") as f:
        json.dump(batch_events, f)
        
    print(f"[PRODUCER] Dispatched batch {batch_num:02d} ({len(batch_events)} transactions) -> {target_file}")
    time.sleep(1.5)

print("[SUCCESS] All test batches generated and staged in landing zone.")
```
